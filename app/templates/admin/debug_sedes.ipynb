```python
from app import create_app
from app.models import Sede

app = create_app()
ctx = app.app_context()
ctx.push()

try:
    sedes = Sede.query.order_by(Sede.created_at.desc()).all()
    print(f"Found {len(sedes)} sedes")
    for s in sedes:
        print(f"ID: {s.id}, Name: {s.name}, Created: {s.created_at}")
except Exception as e:
    print(f"Error querying sedes: {e}")
```
